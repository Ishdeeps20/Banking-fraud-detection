```python
def generate_ai_explanation(row):

    reasons = []

    if row['TransactionAmount'] > 500:
        reasons.append(
            "Unusually high transaction amount detected"
        )

    if row['LoginAttempts'] > 3:
        reasons.append(
            "Multiple login attempts observed"
        )

    if row['LateNightTransaction'] == 1:
        reasons.append(
            "Transaction occurred during late-night hours"
        )

    if row['AccountBalance'] < 100:
        reasons.append(
            "Low account balance before transaction"
        )

    if len(reasons) == 0:
        return "Transaction shows moderate anomaly patterns."

    return ". ".join(reasons)
```

```python
def risk_level(row):

    score = 0

    if row['TransactionAmount'] > 500:
        score += 2

    if row['LoginAttempts'] > 3:
        score += 2

    if row['LateNightTransaction'] == 1:
        score += 1

    if score >= 4:
        return "High Risk"

    elif score >= 2:
        return "Medium Risk"

    else:
        return "Low Risk"
```

```python
fraud_transactions['RiskLevel'] = fraud_transactions.apply(
    risk_level,
    axis=1
)
```

```python
fraud_transactions['AI_Explanation'] = fraud_transactions.apply(
    generate_ai_explanation,
    axis=1
)
```

```python
fraud_transactions['AI_Explanation'] = fraud_transactions.apply(
    generate_ai_explanation,
    axis=1
)
```

```python
import streamlit as st
import pandas as pd
import plotly.express as px

# Load datasets
df = pd.read_csv("full_transactions.csv")
fraud_df = pd.read_csv("fraud_detection_report.csv")

# Title
st.title("AI-Powered Fraud Detection Dashboard")

# Overview
st.subheader("Transaction Overview")

col1, col2 = st.columns(2)

with col1:
    st.metric("Total Transactions", len(df))

with col2:
    st.metric("Fraud Transactions", len(fraud_df))

# Interactive Plotly Scatter Plot
st.subheader("Transaction Amount vs Account Balance")

fig = px.scatter(
    df,
    x="TransactionAmount",
    y="AccountBalance",
    color=df["Anomaly"].astype(str),
    color_discrete_map={
        "-1": "red",
        "1": "blue"
    },
    hover_data=[
        "TransactionAmount",
        "AccountBalance",
        "LoginAttempts"
    ],
    title="Fraud Detection Scatter Plot"
)

st.plotly_chart(fig, use_container_width=True)

# Fraud Transactions Table
st.subheader("Fraud Transactions")

st.dataframe(fraud_df.head(10))

# AI Explanations
st.subheader("AI Fraud Explanations")

if "AI_Explanation" in fraud_df.columns:
    st.dataframe(
        fraud_df[
            [
                "TransactionAmount",
                "AccountBalance",
                "AI_Explanation"
            ]
        ].head(5)
    )
```

```python
import gradio as gr

def check_transaction(amount, balance, attempts):

    if amount > 500 and attempts > 3:
        return "High Fraud Risk"

    elif amount > 300:
        return "Medium Fraud Risk"

    else:
        return "Low Fraud Risk"

demo = gr.Interface(
    fn=check_transaction,
    inputs=[
        gr.Number(label="Transaction Amount"),
        gr.Number(label="Account Balance"),
        gr.Number(label="Login Attempts")
    ],
    outputs="text",
    title="AI Fraud Detection Checker"
)

demo.launch()
```
